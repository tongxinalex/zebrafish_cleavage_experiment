```python
os.chdir(r"H:\Xin home drive\data_meroblastic\Laser_cutter\\laser_cutter_ii\xin250417_lcii_carhoa\\")
post_cut = 8

for i in range(1, 80):
    print('\ncut'+str(i))
    try:
        file = "cut" + str(i) + ".tif"
        data = skimage.io.imread(file)
        fs, ys, xs = data.shape

        plt.imshow(data[post_cut])
        plt.show()

        if 1:
            midy = round(ys/2)
            kymo = data[:, midy-5:midy+5, :]
            kymo = np.average(kymo, axis=1).astype('uint16')
        if 0:
            midx = round(xs/2)
            kymo = data[:, :, midx-5:midx+5]
            kymo = np.average(kymo, axis=2).astype('uint16')
        plt.imshow(kymo)
        plt.show()
        skimage.io.imsave("cut" + str(i) + "_kymo.tif", kymo)
    except:
        print('No such file!')
```
