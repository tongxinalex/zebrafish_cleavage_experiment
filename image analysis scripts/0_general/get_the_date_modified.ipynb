```python
import os, time, scipy, math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
image_size_in_pixel = [2048, 2048]
directory = r"H:/Xin home drive/RAW DATA/Nikon CSU1/xin250710_lcii_wt/"

image_size_per_frame = image_size_in_pixel[0] * image_size_in_pixel[1] * 2

os.chdir(directory)

df = {'file_name':[],
      'date_modified':[],
      'frame_num': []}

for i in os.listdir():
    if '.nd2' in i:
        modification_time = os.path.getmtime(i)
        hh, mm, ss = time.localtime(modification_time)[3:6]
        df['file_name'].append(i)
        df['date_modified'].append(hh*3600 + mm*60 + ss)

        file_size = os.path.getsize(i)
        frame_num = file_size / image_size_per_frame
        df['frame_num'].append(math.floor(frame_num))
        

df = pd.DataFrame(df).sort_values(by = 'date_modified').reset_index(drop='True')
display(df)

df.to_excel('raw_data_info.xlsx')
```
